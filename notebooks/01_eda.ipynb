## 3. Data quality

Missing values, duplicates, numbers stored as text, dates stored as strings. Write down what you find — this is the to-do list for notebook 02.

```python
def data_quality_check(df, name="DataFrame"):
    print(f"\n{'='*60}")
    print(f"DATA QUALITY CHECK: {name}")
    print(f"{'='*60}")

    # 1. Shape
    print("\nShape:")
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    # 2. Data types
    print("\nData types:")
    print(df.dtypes)

    # 3. Missing values
    print("\nMissing values:")
    missing = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_percent": (df.isna().mean() * 100).round(2)
    })
    print(missing[missing["missing_count"] > 0])

    # 4. Duplicate rows
    print("\nDuplicate rows:")
    print(df.duplicated().sum())

    # 5. Number of unique values
    print("\nUnique values per column:")
    print(df.nunique())

    # 6. Detect numeric-looking values stored as text
    print("\nPossible numbers stored as text:")
    
    for col in df.select_dtypes(include="object").columns:
        converted = pd.to_numeric(df[col], errors="coerce")

        original_non_null = df[col].notna().sum()
        numeric_non_null = converted.notna().sum()

        if original_non_null > 0:
            numeric_ratio = numeric_non_null / original_non_null

            if numeric_ratio > 0.5:
                print(
                    f"{col}: {numeric_non_null}/{original_non_null} "
                    f"values look numeric ({numeric_ratio:.1%})"
                )

    # 7. Detect date-like columns stored as strings
    print("\nPossible dates stored as strings:")

    for col in df.select_dtypes(include="object").columns:
        # only test columns whose names suggest dates/years
        if any(word in col.lower() for word in ["date", "year", "time"]):
            converted_dates = pd.to_datetime(
                df[col],
                errors="coerce"
            )

            valid_dates = converted_dates.notna().sum()
            original_non_null = df[col].notna().sum()

            if original_non_null > 0 and valid_dates > 0:
                print(
                    f"{col}: {valid_dates}/{original_non_null} "
                    "values can be interpreted as dates"
                )

    print("\nDone.")
```

```python
data_quality_check(data, "World Bank Data")
```

```python
data_quality_check(countries, "Countries")
```

```python
data_quality_check(indicators, "Indicators")
```
